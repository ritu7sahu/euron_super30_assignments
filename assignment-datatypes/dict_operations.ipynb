```python
laptop = {

"brand": "Dell",

"model": "XPS 15",

"price": 120000,

"ram": "16GB",

"storage": "512GB SSD",

"available": True

}
laptop
```

```python
#print price
laptop.get('price')
```

```python
#Add GPU information
laptop['GPU'] = 'Graphics 360'
laptop
```

```python
#Change RAM to "32GB"
laptop['ram'] ='32GB'
laptop
```
