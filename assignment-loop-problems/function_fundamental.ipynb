```python
"""Create a function that returns the largest of three numbers without using max()."""

def largestNum(lst):
    largestno=[]
    lst.sort()
    lst.reverse()
    largestno.append([lst[0],lst[1],lst[3]])
    return largestno

largestNum([8,1,5,2,9,7,4])

```

```python
"""Create a function that calculates factorial."""

def factorial(num):
    if num == 0 or num == 1:
        return 1
    fact =  num * factorial(num-1) 
    return fact

factorial(5)
    
```

```python
"""Create a function that checks whether a number is prime."""

def checkPrime(num):
    if num <2:
        return "Not a prime Number"
    else:
        for i in range(2,num): 
            if num%i != 0:
                return "No is prime"
            else:
                return "No is not prime"

checkPrime(7)
```

```python
"""Create calculate discount
where the default discount is 10%."""
def calculate_discount(price, discount=10):
    discount_amount = price * discount / 100
    final_price = price - discount_amount
    return final_price

print(calculate_discount(1000))
print(calculate_discount(1000, 20))
#calculate_discount(1000)



```

```python
"""Create a function that accepts a list and returns its sum without using sum()."""

def list_sum(l):
    sum = 0
    for i in l:
        sum += i
    return sum

list_sum([1,2,3,4,5])

```

```python
"""Create a function that accepts a string and returns the number of vowels."""
def countVowel(str):
    vowel = ['a','e','i','o','u']
    count = 0
    for i in str.lower():
        if vowel.count(i) >=1:
            count = count+1
    return count

countVowel("I am here to teach you")


```

```python
"""Create a function that accepts a string and determines whether it is a palindrome."""

def getPallindrom(str):
    left = 0
    right = len(str)-1

    while left < right:
        if str[left]  == str[right]:
           return 'String is Pallindrom'
        else:
            return 'Not Pallindrom'
        
        left +=1
        right -=1

getPallindrom('madam1')



```
