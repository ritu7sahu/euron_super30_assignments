```python
#Print numbers from 1–100 but skip numbers divisible by 5 using continue.

for i in range(1,101):
    if i%5==0:   
        continue
    print(i)
```

```python
#Iterate from 1–100 and stop when you encounter the first number divisible by both 7 and 11.

for i in range(1,101):
    if i%7==0 and i%11==0:
        break

    print(i)
```

```python
#search for a user-provided number inside a list. Use for-else to print:
#"Number Found"
#"Number Not Found"

list1 = [20,10,30,40,90,50,60]
number = int(input('Enter a number'))

for i in list1:
    if number == i:
        print('Number found')
        break
else:
    print('Number not found')
```

```python
names = ["Aman", "Ravi", "Sudhanshu", "Priya", "Anjali"]
# use enumerate() to display
# 1 Aman
# 2 Ravi
# 3 Sudhanshu

for index, value in enumerate(names,start=1):
    print(f"{index}  {value}")

```

```python
#Print the following pattern
# *
# **
# ***
# ****
# *****

for i in range(1,6):
    for j in range(1,i+1):  
        print('*',end="")
    print()
```

```python
# Print
# *****
# ****
# ***
# **     
# *

for i in range(6,0,-1):    
    for j in range(i-1,0,-1):
        print('*',end="")
    print()
```

```python
#Generate multiplication tables from 1 to 10 using nested loops.
for i in range(1,11):
    print(f"multiplication of {i}")   
    for j in range(1,11):
        print(f"{i}*{j} = {i*j} ",end="")
    print()
```

```python
#Find all numbers between 1 and 200 divisible by both 3 and 5.
for i in range(1,201):
    if i%3 == 0 and i%5 == 0:
        print(i)
```

```python
#given a list containing duplicate elements, create another list containing only unique elements without using set().
list2 = [20,10,10,30,20,60,30,20,50,50,60,70]
unique_list = []

for i in list2:
    if unique_list.count(i) >=1:
        continue
    else:
        unique_list.append(i)

unique_list
```

```python
numbers = [10, -4, 8, -2, 0, 15, -9, 21]
#count : positive numbers,negative numbers,zeros
neg_no = 0
pos_no = 0
zero= 0

for i in numbers:
    if i==0:
        zero=zero+1;
    elif i>0:
        pos_no+=1
    elif i<0:
        neg_no+=1

print(f"count of positive number : {pos_no}")
print(f"count of negative number : {neg_no}")
print(f"count of zero :  {zero}")
```

```python
#Write a program to determine whether a number is prime using a loop.
num1 = int(input('Enter a number'))
if(num1<2):
    print('Number is prime no')
else:
    for i in range(2,num1):
        if num1 % i == 0:
            print('Not a prime number')
            break
        else:
            print('Prime number')
            break;
```
