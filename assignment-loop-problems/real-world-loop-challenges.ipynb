```python
marks = [78, 92, 45, 67, 88, 53, 99]

#count how many students scored 90+, 75–89, 50–74, below 50
countA = 0
countB=0
countC=0
countD=0
for i in marks:
    if i >= 90:
        countA+=1
    elif i>75 and i<90:
        countB+=1
    elif i<75 and i>=50:
        countC+=1
    elif i<50:
        countD+=1

print(f"student scored 90+ {countA}")
print(f"student scored 75-89 {countB}")
print(f"student scored 50-74 {countC}")
print(f"student scored below 50 {countD}")

```

```python
#Create a simple login system with a maximum of 3 password attempts.
username = "RituS"
password = "Ritu1234"

user = input('Enter the username')
count = 0
while count<3:
    pwd = input('Enter the password')
    if user == username and pwd == password:
        print('Login Success')
        break
    elif user == username and pwd != password:
        count = count+1
        print('Enter the correct pwd')
    elif user != username:
        print('Incorrect username')
        break;

if count ==3:
    print('You have exceeded with password attempts')
    

```

```python
#Find the second-largest number in a list without using sort().
trans = [10,30,20,5,50,90,60]
first = 0
second = 0

for i in trans:
    if i> first:    
        second = first
        first = i
    elif i>second and i != first: 
        second = i

print(f"second-largest number : {second}")
```

```python
#Check whether a string is a palindrome using loops.
str2 = 'madam'
left = 0
right = len(str2)-1

while left < right:
    if str2[left]  == str2[right]:
        print('String is Pallindrom')
        break
    else:
        print('Not Pallindrom')
        break
    
    left +=1
    right -=1

```

```python
#Create a basic ATM simulation where a user can repeatedly
# Check balance
# Deposit money
# Withdraw money
# Exit

balance = 10000
while True:
    print("\n Welcome to ATM")
    print("1. Check Balance")
    print("2. Deposite Money")
    print("3. Withdraw Money")
    print("4. Exit")

    choice = input("Enter your Choice")
    if choice == "1":
        print(f"Your account balance is {balance}")
        break
    elif choice == "2":
        amount = int(input("Enter the amount"))
        if amount>0:
            balance = amount+balance
            print("Deposite successfully")
            print(f"Now your balance is: {balance}")
            break
        else:
            print('Enter valid amount')
            break;
    elif choice == "3":
        amount = int(input("Enter the amount"))
        if(amount>0):
            balance = balance-amount;
            print("Amount Withdrom successfully")
            print(f"Your account balance is {balance}")
            break
        else:
            print("Enter a valid amount")
            break
    elif choice == '4':
        print("Thanks for choosing ATM service")
        break
    else:
        print("Enter a valid choice")
        break
```
